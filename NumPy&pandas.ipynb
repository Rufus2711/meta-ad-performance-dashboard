pandas Multi-Condition Filtering

Display ONLY names where:

Country is India
AND
Sales > 5000
OR
Experience >= 5

```python
import pandas as pd

data = {
    "Name":["Rufus","Amit","Sara","John"],
    "Sales":[7000,3000,9000,4000],
    "Country":["India","India","USA","India"],
    "Experience":[3,1,5,2]
}

df=pd.DataFrame(data)

df[((df["Country"]=="India") & (df["Sales"]>5000)) | (df["Experience"]>=5)]["Name"]
```

Task

Create new column:Level


Rules:

Sales >= 8000 → "Senior"
Sales >= 5000 → "Mid"
otherwise → "Junior"

Use:

apply()
lambda

```python
df["Level"]=df["Sales"].apply (
    lambda x: "Senior" if x >=8000
    else "Mid" if x>=5000 else "Junior"
)

df["Level"]
```
